```sql
%python

# F String [Dynamic Variable passing]

input_path = "/Volumes/dev/demo/raw/drivers.json"
catalog = "dev"
schema = "demo"
table_name = "drivers_RW_API"
```

```sql
%python
( # reader API
spark
.read
.json(f"{input_path}")
# Writer API
.write
.mode("overwrite")
.saveAsTable(f"{catalog}.{schema}.{table_name}")
)
```
