CTAs - Create table as select ||
read_file - help you read any form of file [csv, parquet, json, txt etc.]


```sql
create table if not exists dev.demo.circuits_sql as 
select *, _metadata.file_path, _metadata.file_name, current_timestamp() as ingested_time from read_files("/Volumes/dev/demo/raw/circuits.csv")
```

```sql
SELECT * FROM dev.demo.circuits_sql
```
