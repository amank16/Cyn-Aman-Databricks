```sql
create view if not exists dev.demo.order_priority as
SELECT 
o_orderpriority as Order_Priority, 
count(*) as Count 
FROM samples.tpch.orders 
GROUP BY o_orderpriority 
ORDER BY Count DESC
```

```sql
create function if not exists dev.demo.datamask(str string)
returns string 
return 
concat(
    substr(str, 1, 1),
    repeat("*", length(str)-2),
    substr(str, -1, 1)
);
```

```sql
SELECT *, dev.demo.datamask(email_address) as masked_email FROM samples.bakehouse.sales_customers
```
